### 1-1

```python
september_expense = [300000, 50000, 70000, 30000]

print(september_expense)
```

### 1-2

```python
september_expense.append(400000)

september_expense[1] = 60000

print(september_expense)
```

### 1-3

```python
del september_expense[3]

print(september_expense)
```

### 1-4

```python
print(30000 in september_expense)

print(len(september_expense))
```

### 1-5

```python
september_expense.sort()

print(september_expense)

september_expense.sort(reverse=True)

print(september_expense)
```

### 1-6

```python
september_expense += [150000]
september_expense += [80000]

print(september_expense)
```
