### 1-1


```python
def fahr2kelvin(fahr):
    celsius = (float(fahr) - 32.0) / 1.8
    return celsius + 273.15

print(fahr2kelvin(10))
```

### 1-2


```python
def celsius2fahr(celsius):
    return celsius * 1.8 + 32

print(celsius2fahr(10))
```

### 1-3


```python
def cel2FK(celsius):
    fahr = celsius * 1.8 + 32
    kelvin = celsius + 273.15

    return {"fahr": fahr, "kelvin": kelvin}

print(cel2FK(10))
```

### 2-1


```python
def factorial(num):
    if num <= 1:
        return 1
    return num * factorial(num - 1)

def bday(num):
    if num > 365:
        return

    return 1 - factorial(365) / (factorial(365 - num) * 365**num)

print(bday(2))
```
