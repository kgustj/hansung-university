```python
import pandas as pd

houseSales = pd.read_csv("house_sales.csv", encoding="utf-8")
houseInfo = pd.read_csv("house_basicinfo.csv", encoding="utf-8")
aptSales = pd.read_csv("apt_sales.csv", encoding="utf-8")
aptBasicInfo = pd.read_csv("apt_basicinfo.csv", encoding="utf-8")
```

### 1-1


```python
print(houseSales.shape[1])
houseSales.columns

# 총 8개의 열, salesPrice는 8번째 열
```

### 1-2


```python
house1 = pd.merge(houseSales, houseInfo, on="id", how="left")
print(house1)
```

### 1-3


```python
house2 = house1.drop(columns=["si_y", "gu_y", "dong_y", "bunji_y"])
print(house2)
```

### 1-4


```python
house = house2.rename(
    columns={
        "si_x": "si",
        "gu_x": "gu",
        "dong_x": "dong",
        "bunji_x": "bunji",
    }
)
print(house)
```

### 1-5


```python
ddm = house[house["gu"] == "동대문구"]
(ddm["salesPrice"] / ddm["landSize"]).describe()
```

### 1-6


```python
print(house[house["gu"] == "동대문구"]["salesPrice"].mean())
print(house[house["gu"] == "성북구"]["salesPrice"].mean())
```

### 2-1


```python
print(aptSales.shape[1])
aptSales.columns

# 총 10개의 열, salesPrice는 10번째 열
```

### 2-2


```python
apt1 = pd.merge(aptSales, aptBasicInfo, on="id", how="left")
print(apt1)
```

### 2-3


```python
apt2 = apt1.drop(columns=["si_y", "gu_y", "dong_y", "bunji_y", "mainBun_y", "subBun_y"])
print(apt2)
```

### 2-4


```python
apt = apt2.rename(
    columns={
        "si_x": "si",
        "gu_x": "gu",
        "dong_x": "dong",
        "bunji_x": "bunji",
        "mainBun_x": "mainBun",
        "subBun_x": "subBun",
    }
)
print(apt)
```

### 2-5


```python
sb = apt[apt["gu"] == "성북구"]
(sb["salesPrice"] / sb["size"]).describe()
```

### 2-6


```python
davg = apt[apt["gu"] == "동대문구"]["salesPrice"].mean()
savg = apt[apt["gu"] == "성북구"]["salesPrice"].mean()

apt["guAvgPrice"] = apt["gu"].map({"동대문구": davg, "성북구": savg})
apt[["gu", "salesPrice", "guAvgPrice"]]
```
